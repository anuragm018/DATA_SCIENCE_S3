```python
#PRGM -3
#Q--1

import matplotlib.pyplot as plt

x = [1, 2, 6, 18]
y = [3, 10, 12, 20]

plt.plot(x, y, linestyle=':', color='blue', marker='o',
         markerfacecolor='red', markeredgecolor='red')

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
#Q-2

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 8, 27, 64, 125]

plt.subplot(1, 2, 1)
plt.plot(x, y1, color='blue')
plt.title("Square")

plt.subplot(1, 2, 2)
plt.plot(x, y2, color='green')
plt.title("Cube")

plt.tight_layout()
plt.show()
```

```python
#Q--3

import matplotlib.pyplot as plt
import numpy as np

men = [22, 30, 35, 35, 26]
women = [25, 32, 30, 35, 29]

x = np.arange(5)
width = 0.35

plt.bar(x - width/2, men, width, label='Men', color='blue')
plt.bar(x + width/2, women, width, label='Women', color='pink')

plt.xticks(x, ['G1', 'G2', 'G3', 'G4', 'G5'])
plt.xlabel("Groups")
plt.ylabel("Scores")
plt.title("Scores by Group and Gender")
plt.legend()
plt.show()
```

```python
#Q -4
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 4, 9, 16, 25]

plt.plot(x, y1, label='Line 1', color='blue')
plt.plot(x, y2, label='Line 2', color='red')

plt.title("Multiple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#q-5
import matplotlib.pyplot as plt

languages = ['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8.0, 7.7, 6.7]

# Pie Chart

plt.pie(popularity, labels=languages, autopct='%1.1f%%')
plt.title("Programming Language Popularity")
plt.show()

# Scatter Plot

plt.scatter(languages, popularity, color='red')
plt.title("Scatter Plot")
plt.xlabel("Languages")
plt.ylabel("Popularity")
plt.grid(True)
plt.show()

# Horizontal Bar Chart

plt.barh(languages, popularity, color='green')
plt.title("Horizontal Bar Chart")
plt.xlabel("Popularity")
plt.grid(True)
plt.show()
```
